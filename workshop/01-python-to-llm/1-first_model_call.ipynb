# 01 — From a Python function to a model call
Problem: fixed rules cannot synthesize an answer when evidence is incomplete.
Watch what changes when the program calls a model: the output becomes useful,
but the model may assert something that the input never established.



## Problem

A fixed Python rule can compare numbers but cannot explain uncertain evidence.


## Concept

A model call adds synthesis while making factual verification necessary.

A Python function maps explicit inputs to a repeatable output. An LLM can synthesize language across incomplete inputs, but its response is probabilistic and must be checked against the evidence supplied.


## Abstraction

facts → Python function / CrewAI LLM → compare outputs and unknowns. This is the smallest new layer introduced by this lesson; the earlier layers remain in place.


## Architecture

```text
same evidence ──┬──▶ fixed Python rule ──▶ fixed decision
               └──▶ CrewAI LLM call ──▶ prose + possible unsupported claims
                                      │
                                      └──▶ human checks evidence
```

Follow the arrows while executing the cells. A tool result or validation failure is an observation, not an instruction that the model may bypass.


## Step 1 — Run the deterministic rule with known inputs

Run the deterministic rule with known inputs; note what it cannot infer.


````python
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).resolve().parents[2] if "__file__" in globals() else Path.cwd()))

from workshop.common import model


def fixed_recommendation(annual_benefit: int, annual_cost: int) -> str:
    return "pilot" if annual_benefit > annual_cost else "investigate further"


print("Python rule:", fixed_recommendation(45_000, 30_000))

````

## The first model call
CrewAI's LLM object supplies the model here. The input deliberately omits
security evidence. Compare the model response with the facts provided.



## Step 2 — Call the CrewAI LLM on the same facts and mark any assertion unsupported by the prompt

Call the CrewAI LLM on the same facts and mark any assertion unsupported by the prompt.


````python
response = model().call(
    "A fictional AI platform may save $45,000/year and cost $30,000/year. "
    "We have not verified data deletion or throughput. Recommend adopt, pilot, "
    "or reject in two sentences. State what is still unknown."
)
print("Model response:", response)
````

## Execution flow

facts → Python function / CrewAI LLM → compare outputs and unknowns. Read the printed result together with the task, selected tool, state, or trace. If the result differs, locate the first boundary where the observed value changed.


## Failure mode to know about

The model may present the missing deletion guarantee as a fact. A fluent answer is not evidence; write down which input sentence supports each claim.


## Production implication

Use the model for synthesis, not as a source of truth. Keep arithmetic and known business rules in deterministic code.
